Write read_csv(path) returning a list of dictionaries.

```python
items = read_csv(f"{data_folder}/olist_order_items_dataset.csv")
orders = read_csv(f"{data_folder}/olist_orders_dataset.csv")
validate_rows(items, ["order_id", "price", "freight_value"])
items[:2]
```

Write a function that converts column names to snake_case.

```python
items = normalize_columns(items)
orders = normalize_columns(orders)
items[:2]
```

Load the column mapping from a JSON config file instead of code.

```python
mapping = read_mapping("column_mapping.json")
```

Create a column-mapping dictionary to a standard schema.

```python
items = map_columns(items, mapping)
orders = map_columns(orders, mapping)
items[:2]
```

Convert price and freight to float, and timestamps to datetime.

```python
items = convert_values(items)
orders = convert_values(orders)
items[:2], orders[:2]
```

```python
order_totals = calculate_order_totals(items)
list(order_totals.items())[:5]
```
