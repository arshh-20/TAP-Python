Read olist_orders_dataset.csv with the csv module and print the row count

```python
import csv

with open(f"{input_folder}/olist_orders_dataset.csv", newline="", encoding="utf-8") as file:
    row_count = sum(1 for row in csv.reader(file)) - 1

print("Row count:", row_count)
```

Loop over all Olist CSV files and print a summary for each.

```python
import glob

for path in glob.glob(f"{input_folder}/*.csv"):
    data = pd.read_csv(path, sep=None, engine="python")
    print(path,"\n")
    data.info()
```

Make a copy of one file with ';' as the delimiter and handle both formats.

```python
source = f"{input_folder}/olist_customers_dataset.csv"
copy_path = f"{input_folder}/olist_customers_semicolon.csv"

customers = pd.read_csv(source)
customers.to_csv(copy_path, sep=";", index=False)

semicolon_customers = pd.read_csv(copy_path, sep=";")

print("Semicolon copy rows:", len(semicolon_customers))
print("Semicolon copy columns:", semicolon_customers.columns.tolist())
```
